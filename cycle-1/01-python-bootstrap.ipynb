```python
# Module 2 Exo C — Composition de fonctions
# Écrire 2 fonctions :
#   - `kilos_to_grams(k)` retourne k * 1000
#   - `grams_to_milligrams(g)` retourne g * 1000
# Écrire une 3e `kilos_to_milligrams(k)` qui utilise les 2 précédentes
# en interne (pas faire k * 1_000_000 directement). C'est l'idée de
# composer des fonctions.
# Tester avec kilos_to_milligrams(2) (attendu : 2_000_000).

def grams_to_milligrams(g) : 
    return g * 1000

def kilos_to_grams(k) : 
    return k * 1000

def kilos_to_milligrams(k) : 
    grams = kilos_to_grams(k)
    return grams_to_milligrams(grams)



kilos_to_milligrams(2)
```

```python
# Module 3 Exo 1 — Rachat de dette : refaire `tier` proprement
# Avec return + elif cette fois.
# tier(20) → "T1", tier(50) → "T2", tier(80) → "T3"

def tier(score):
    if score < 30:
        return "T1"
    elif score <= 70:
        return "T2"
    else:
        return "T3"

# Tests sur les 3 cas de l'énoncé
print(tier(20))   # T1
print(tier(50))   # T2
print(tier(80))   # T3
```

```python
# Module 3 Exo 2 — Mini-fonction utile : mention(score)
# Prend un score sur 100, retourne une mention :
#   - score < 50  → "Insuffisant"
#   - score < 60  → "Passable"
#   - score < 70  → "Assez bien"
#   - score < 80  → "Bien"
#   - score >= 80 → "Très bien"
# Doit gérer un score invalide (négatif ou > 100) → retourner "Invalide"

def mention(score):
    if score < 0 or score > 100:
        return "Invalide"
    elif score < 50:
        return "Insuffisant"
    elif score < 60:
        return "Passable"
    elif score < 70:
        return "Assez bien"
    elif score < 80:
        return "Bien"
    else:
        return "Très bien"

mention(100)
```

```python
with open("lexique.md", "r", encoding="utf-8") as f:
    contenu = f.read()
def word_counter_veille(note) :
    
    STOPWORDS = {"|", "-", ":", "de", "✅", "—", "je", "le", "=", "en", "la", "+",
             "j'ai", "---", "##", "un", "une", "des", "les", "et", "à", "dans",
             "sur", "ni", "→", "déjà", "où", "pas", "si", "1", "2", "in", "for", "pour", "par", "avec", "3", "5", "0", "/", "x", "a", "b", "if", "n", "qui", "[", "]", "⭐"}
    
    counts = {}
    words = note.split()
    for word in words :
        word = word.strip("`.,()*#:!?'\"—").lower()
        if word and word not in STOPWORDS:    # skip les strings vides après strip
            word = word.lower()
            counts[word] = counts.get(word, 0) + 1
    tri = sorted(counts.items(), key=lambda kv: kv[1], reverse=True)
    final = tri[:10]
    return final    
    
top10 = word_counter_veille(contenu)
print(top10)
```

```python
class Note:
    def __init__(self, title, theme):
        # self = l'objet en cours de construction, comme `this` en JS
        # On y "accroche" les attributs avec self.attribut = valeur
        self.title = title
        self.theme = theme
        self.tags = []                # valeur par défaut

    def add_tag(self, tag):           # méthode = fonction qui prend self
        self.tags.append(tag)

    def display(self):
        print(f"[{self.theme}] {self.title} ({(self.tags)} tags)")

# Utilisation
n = Note("ML basics", "ai-llms")      # appelle __init__ (pas .new(), pas de mot-clé `new`)
n.add_tag("evergreen")
n.add_tag("to-process")
n.display()                           # [ai-llms] ML basics (2 tags)
```

```python
# Crée class Note avec __init__ prenant title, theme, tier (par défaut "T2")
# Ajoute une méthode set_rating(self, rating) qui stocke self.rating = rating
# Ajoute une méthode summary(self) qui retourne (pas print) une string genre "ML basics [ai-llms · T2 · ⭐4]"
# Instancie 2 notes différentes, set le rating de l'une, et print summary() des deux
class Note:
    def __init__(self, title, theme, tier="T2"):
        self.title = title
        self.theme = theme
        self.tier = tier
        self.rating = None             

    def set_rating(self, rating):
        self.rating = int(rating)      

    def summary(self):
        rating_str = f"⭐{self.rating}" if self.rating is not None else "⭐?"
        return f"{self.title} [{self.theme} · {self.tier} · {rating_str}]"

n1 = Note("ML basics", "ai-llms", "T2")
n2 = Note("Numpy intro", "dev-code")          
n1.set_rating(4)                              

print(n1.summary())  
print(n2.summary())   
```

```python
# Écris class VideoNote(Note) qui hérite de Note
# __init__ prend title, theme, duration_sec (entier en secondes), platform (str genre "youtube", "tiktok"), et tier="T2" par défaut
# Utilise super().__init__(...) pour construire la partie Note
# Override summary() pour ajouter durée formatée + plateforme — format libre, du genre : "Karpathy GPT [ai-llms · T2 · ⭐5] (youtube · 117:32)"
# Crée 1 instance de Note (texte) + 1 instance de VideoNote, set_rating les deux, print les summary() des deux

class VideoNote(Note):
    def __init__(self, title, theme, duration_sec, platform, tier="T2"):
        super().__init__(title, theme, tier)
        self.duration_sec = duration_sec
        self.platform = platform
        
    def set_rating(self, rating):
        self.rating = int(rating)    
        
    def summary(self):
        base = super().summary()
        minutes = self.duration_sec // 60
        seconds = self.duration_sec % 60        
        return f"{base} ({self.platform} · {minutes}:{seconds:02d}) "
    
n = Note("Karpathy", "ai-llms",  "T2", )
n1 = VideoNote(n.title, n.theme, 8500, "youtube")
n1.set_rating(5)
n.set_rating(3)

print(n.summary())   
print(n1.summary())  
```
